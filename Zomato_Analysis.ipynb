```python
import pandas as pd

df = pd.read_csv("../Dataset/zomato.csv", encoding="latin1")

country_df = pd.read_excel("../Dataset/country-code.xlsx")

df = df.merge(country_df, on="Country Code", how="left")

cuisine_split = df["Cuisines"].str.split(",", expand=True)

cuisine_split.columns = [
    f"Cuisine_{i+1}" for i in range(cuisine_split.shape[1])
]

cuisine_split = cuisine_split.apply(lambda x: x.str.strip())

df = pd.concat([df, cuisine_split], axis=1)

df.head()
```

```python
df["Aggregate rating"].dtype
```

```python
df["Aggregate rating"].value_counts().sort_index()
```

```python
df["Aggregate rating"] = df["Aggregate rating"].replace(0.0, pd.NA)
```
